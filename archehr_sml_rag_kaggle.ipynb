```python
val_case_ids = set(val_df["case_id"])
validation_cases = [c for c in dev_cases if c.case_id in val_case_ids]

def cited_sentence_ids(answer):
    groups = re.findall(r"\[([^\]]+)\]", answer)
    return {
        item
        for group in groups
        for item in re.findall(r"\d+", group)
    }

def set_prf(predicted, gold):
    predicted, gold = set(predicted), set(gold)
    tp = len(predicted & gold)
    precision = tp / len(predicted) if predicted else 0.0
    recall = tp / len(gold) if gold else 0.0
    f1 = (
        2 * precision * recall / (precision + recall)
        if precision + recall else 0.0
    )
    return precision, recall, f1

validation_results = [answer_case(case) for case in validation_cases]

rows = []
rouge = rouge_scorer.RougeScorer(["rougeL"], use_stemmer=True)

for case, result in zip(validation_cases, validation_results):
    gold_evidence = {
        sentence_id
        for sentence_id, label in case.labels.items()
        if label == "essential"
    }

    retrieved_evidence = {item["id"] for item in result["evidence"]}
    cited_evidence = cited_sentence_ids(result["answer"])

    retrieval_p, retrieval_r, retrieval_f1 = set_prf(
        retrieved_evidence, gold_evidence
    )
    citation_p, citation_r, citation_f1 = set_prf(
        cited_evidence, gold_evidence
    )

    rouge_l = rouge.score(
        case.answer,
        result["answer"]
    )["rougeL"].fmeasure

    rows.append({
        "case_id": case.case_id,
        "retrieval_precision": retrieval_p,
        "retrieval_recall": retrieval_r,
        "retrieval_f1": retrieval_f1,
        "citation_precision": citation_p,
        "citation_recall": citation_r,
        "citation_f1": citation_f1,
        "rougeL": rouge_l,
        "self_consistency_confidence": result["confidence_self_consistency"],
        "reference_answer": case.answer,
        "generated_answer": result["answer"],
        "evidence_trace": result["evidence_trace"],
    })

evaluation_df = pd.DataFrame(rows)

bleu = corpus_bleu(
    evaluation_df["generated_answer"].tolist(),
    [evaluation_df["reference_answer"].tolist()]
).score

print("===== Mean validation metrics =====")
print(evaluation_df[
    [
        "retrieval_precision",
        "retrieval_recall",
        "retrieval_f1",
        "citation_precision",
        "citation_recall",
        "citation_f1",
        "rougeL",
        "self_consistency_confidence",
    ]
].mean())

print(f"\nBLEU: {bleu:.2f}")

display(evaluation_df)

evaluation_df.to_csv(
    "/content/archehr_sml_rag/validation_evaluation.csv",
    index=False,
)
```

```python
sns.set_theme(style="whitegrid")

plot_df = evaluation_df.copy()

plot_df["trace_claim_count"] = plot_df["evidence_trace"].apply(len)
plot_df["avg_evidence_per_claim"] = plot_df["evidence_trace"].apply(
    lambda trace: (
        sum(len(item["supporting_sentence_ids"]) for item in trace) / len(trace)
        if len(trace) else 0
    )
)

fig, axes = plt.subplots(2, 2, figsize=(18, 12))

mean_metrics = plot_df[
    [
        "retrieval_precision",
        "retrieval_recall",
        "retrieval_f1",
        "citation_precision",
        "citation_recall",
        "citation_f1",
        "rougeL",
        "self_consistency_confidence",
    ]
].mean()

sns.barplot(
    x=mean_metrics.index,
    y=mean_metrics.values,
    ax=axes[0, 0],
    color="#4C78A8",
)
axes[0, 0].set_title("Mean validation metrics")
axes[0, 0].set_ylabel("Score")
axes[0, 0].set_xlabel("")
axes[0, 0].set_ylim(0, 1)
axes[0, 0].tick_params(axis="x", rotation=45)

f1_df = plot_df.melt(
    id_vars="case_id",
    value_vars=["retrieval_f1", "citation_f1"],
    var_name="metric",
    value_name="score",
)

sns.barplot(
    data=f1_df,
    x="case_id",
    y="score",
    hue="metric",
    ax=axes[0, 1],
)
axes[0, 1].set_title("Evidence retrieval vs. answer citation F1")
axes[0, 1].set_ylabel("F1 score")
axes[0, 1].set_xlabel("Validation case")
axes[0, 1].set_ylim(0, 1)

generation_df = plot_df.melt(
    id_vars="case_id",
    value_vars=["rougeL", "self_consistency_confidence"],
    var_name="metric",
    value_name="score",
)

sns.barplot(
    data=generation_df,
    x="case_id",
    y="score",
    hue="metric",
    ax=axes[1, 0],
)
axes[1, 0].set_title("Generated-answer quality and self-consistency")
axes[1, 0].set_ylabel("Score")
axes[1, 0].set_xlabel("Validation case")
axes[1, 0].set_ylim(0, 1)

trace_df = plot_df.melt(
    id_vars="case_id",
    value_vars=["trace_claim_count", "avg_evidence_per_claim"],
    var_name="metric",
    value_name="value",
)

sns.barplot(
    data=trace_df,
    x="case_id",
    y="value",
    hue="metric",
    ax=axes[1, 1],
)
axes[1, 1].set_title("Evidence-trace coverage")
axes[1, 1].set_ylabel("Count")
axes[1, 1].set_xlabel("Validation case")

plt.tight_layout()
plt.show()
```
